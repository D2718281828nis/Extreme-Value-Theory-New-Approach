# 4. EVT и AR baseline

Сохраняем отображение точечных срабатываний и добавляем отдельный график качества. Метрики считаются относительно известного активного интервала синтетического события.

```python
import matplotlib.pyplot as plt
from evt_demo import generate_graph_time_series
from evt_demo.pipeline import prepare_sample
from evt_demo.evaluation import compare_evt_ar
from evt_demo.visualization import plot_detection_quality
series=generate_graph_time_series(seed=42)
_,detection=prepare_sample(series)
comparison=compare_evt_ar(series,detection)
fig,ax=plt.subplots(figsize=(12,4)); ax.plot(detection.indicator,color="0.6",label="индикатор"); ax.scatter(__import__("numpy").flatnonzero(comparison.evt_mask),detection.indicator[comparison.evt_mask],label="EVT"); ax.scatter(__import__("numpy").flatnonzero(comparison.ar_mask),detection.indicator[comparison.ar_mask],label="AR"); ax.axvspan(series.event_onset,series.arrival_times.max()+18,color="tab:red",alpha=.12,label="истинный интервал"); ax.legend(); ax.set_title("Отображение срабатываний EVT и AR");
```

```python
plot_detection_quality(comparison);
{n:{"precision":float(p),"recall":float(r),"f1":float(f)} for n,p,r,f in zip(comparison.names,comparison.precision,comparison.recall,comparison.f1)}
```

Первый рисунок отвечает на вопрос «где сработал метод», второй — «насколько срабатывания совпадают с истиной». Результат зависит от определения интервала и порогов.
