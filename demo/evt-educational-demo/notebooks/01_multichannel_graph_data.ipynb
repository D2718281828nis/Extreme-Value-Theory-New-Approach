Генерируем воспроизводимый AR(1)-фон и экстремальное событие, распространяющееся от одного узла.

```python
# Локальный пакет: эта настройка находится в той же ячейке, что и импорт API.
from pathlib import Path
import importlib
import sys

cwd = Path.cwd().resolve()
candidates = [cwd / "src", cwd.parent / "src", cwd / "demo" / "evt-educational-demo" / "src"]
src = next((p for p in candidates if (p / "evt_demo" / "__init__.py").exists()), None)
if src is None:
    raise RuntimeError(f"Не найден evt_demo/src из {cwd}. Откройте репозиторий целиком в VS Code.")
sys.path.insert(0, str(src))
for name in [n for n in sys.modules if n == "evt_demo" or n.startswith("evt_demo.")]:
    del sys.modules[name]
importlib.invalidate_caches()
print(f"evt_demo: {src}")

from evt_demo import generate_graph_time_series
series = generate_graph_time_series(seed=42)
series.values.shape, series.source_node, series.event_onset
```
